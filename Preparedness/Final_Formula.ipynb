```python
import numpy as np
from scipy.stats import gmean
import math

def min_max_rescale(minimum, maximum, value):
    if maximum == minimum:
        return value
    return (value - minimum) / (maximum - minimum)

def preparedness_index(df, epsilon=0.1, weights=(0.33, 0.33, 0.34), cap=10):
    n_hazard_maps = np.log(10.0 * df["n_hazard_maps"]/math.log(100.0)).clip(upper=1).fillna(0)
    rescaled_health_per_capita = min_max_rescale(df["n_health_per_capita"].min(), df["n_health_per_capita"].max(), df["n_health_per_capita"])
    rescaled_HDI = min_max_rescale(df["HDI"].min(), df["HDI"].max(), df["HDI"]).fillna(df["HDI"].mean())
    comps = np.column_stack([n_hazard_maps, rescaled_health_per_capita, rescaled_HDI]).clip(epsilon, 1)
    return gmean(comps, axis=1, weights=weights)
```

```python
import pandas as pd
df = pd.read_csv("./volcano_preparedness_data.csv")
```

```python
df["n_health_per_capita"] = 5.0
```

```python
df["Preparedness Index"] = preparedness_index(df)
```

```python
from load_gvp import load_gvp
import pandas as pd

volcanoes = load_gvp("./Databases/GVP_Volcano_List_Holocene_202610190409.xls")
```

```python
volcanoes
```

```python
country_code_df = pd.read_csv("./Databases/country-codes.csv")
country_code_df = country_code_df[["ISO3166-1-Alpha-3","official_name_en"]].rename(columns={"official_name_en":"Country", "ISO3166-1-Alpha-3":"Country Code"})
volcanoes = pd.merge(volcanoes, country_code_df, on="Country")
volcanoes
```

```python
subnational_HDI = pd.read_excel("./Databases/GDL-Subnational-HDI-data.xlsx")[["ISO_Code",2023,"Level"]].rename(columns={"ISO_Code":"Country Code", 2023:"HDI"})
national_HDI = subnational_HDI[subnational_HDI['Level'] == "National"]
```

```python
national_HDI
```

```python
volcanoes_HDI = pd.merge(volcanoes, national_HDI, on="Country Code")
```

```python
def monitoring_stations_counts(path):
    df = pd.read_excel(path)
    print(df)
    df["End Date"] = pd.to_datetime(df["End Date"], errors="coerce")

    df["is_active"] = df["End Date"].isna() | (df["End Date"] >= pd.Timestamp.today())
 
    g = df.groupby("Volcano Number")
    return pd.DataFrame({
        "n_active_monitoring_stations": df[df["is_active"]].groupby("Volcano Number")["Station Code"].nunique(),
    }).fillna(0).astype(int).reset_index()
```

```python
monitoring_stations_df = monitoring_stations_counts("./Databases/Seismic monitoring system.xlsx")
monitoring_stations_df
```

```python
volcano_HDI_monitoring = pd.merge(volcanoes_HDI, monitoring_stations_df, on='Volcano Number', how="left")
```

```python
volcano_HDI_monitoring
```

```python
maps = pd.read_csv("Number of hazard maps.csv")
```

```python
final_df = pd.merge(volcano_HDI_monitoring, maps, on='Volcano Number', how="left")
```
